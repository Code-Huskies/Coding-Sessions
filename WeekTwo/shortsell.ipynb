```python
import sys
import io

test_data = '''5 10
1000 980 960 940 10'''

test_data = '''5 100
100 100 100 100 100'''


sys.stdin = io.StringIO(test_data)
input = sys.stdin.readline

N, K = tuple([int(e) for e in input().split()])
a = [int(e)*100 for e in input().split()]
```

```python
def build_prefix(a):
    p = [0]*(len(a))
    for k in range(1, len(a)):
        p[k] = a[k - 1] - a[k] - K
    return p
```

```python
print(a)
build_prefix(a)
```

```python
def max_subarray(a):
    best = a[0]           # best total found anywhere
    cur = a[0]            # best total of a run ending here
    lo = hi = 0           # where that best run sits
    start = 0             # where the current run began

    for i in range(1, len(a)):
        if cur < 0:            # the run so far is a net loss
            cur = a[i]         # drop it, start again at i
            start = i
        else:
            cur = cur + a[i]   # keeping it can only help

        if cur > best:         # best run seen so far?
            best = cur
            lo, hi = start, i

    return best, lo, hi


best, _, _ = max_subarray(a)


print(max(0, max_subarray(build_prefix(a))[0] - K))
```
